```python
from queue import PriorityQueue

def a_star(graph, start, goal, heuristic):
    open_set = PriorityQueue()
    # Store g_score along with the node to track the cost when queued
    open_set.put((0 + heuristic[start], 0, start)) 
    came_from = {}
    g_score = {node: float('inf') for node in graph}
    g_score[start] = 0
    
    while not open_set.empty():
        f, current_g, current = open_set.get()
        
        # Stop immediately when the goal is reached
        if current == goal:
            path = []
            while current in came_from:
                path.append(current)
                current = came_from[current]
            path.append(start)
            return path[::-1]
            
        # SKIP if we found a shorter path to this node since it was queued
        if current_g > g_score[current]:
            continue
            
        for neighbor, cost in graph[current]:
            tentative_g = g_score[current] + cost
            if tentative_g < g_score[neighbor]:
                came_from[neighbor] = current
                g_score[neighbor] = tentative_g
                f_score = tentative_g + heuristic[neighbor]
                open_set.put((f_score, tentative_g, neighbor))
                
    return None

# Define graph and heuristic
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 1)],
    'D': [('G', 2)],
    'E': [('G', 1)],
    'F': [('G', 3)],
    'G': []
}

heuristic = {
    'A': 7, 'B': 6, 'C': 2, 'D': 1, 'E': 0, 'F': 3, 'G': 0
}

path = a_star(graph, 'A', 'G', heuristic)
print("Shortest path using A*:", path)
```
