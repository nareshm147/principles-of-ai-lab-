```python
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    visited.add(start)
    print(start, end=" ")

    for neighbor in graph[start]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)



graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

print("DFS traversal from node A:")
dfs(graph, 'A')
```

```python
def minimax(depth, nodeIndex, isMax, scores, h):
    # Base case: reached the leaf node
    if depth == h:
        return scores[nodeIndex]

    # Maximizer's turn
    if isMax:
        return max(
            minimax(depth + 1, nodeIndex * 2, False, scores, h),
            minimax(depth + 1, nodeIndex * 2 + 1, False, scores, h)
        )

    # Minimizer's turn
    else:
        return min(
            minimax(depth + 1, nodeIndex * 2, True, scores, h),
            minimax(depth + 1, nodeIndex * 2 + 1, True, scores, h)
        )


scores = [3, 5, 6, 9, 1, 2, 0, -1]
h = 3

print("Optimal value:", minimax(0, 0, True, scores, h))
```
